```python
documents = {
    1: "Python is a popular programming language.",
    2: "Java is also a popular language for enterprise development.",
    3: "Python is great for data science and machine learning.",
    4: "Data science requires strong programming skills."
}

for doc_id, text in documents.items():
    print(f"Doc {doc_id}: {text}")
```

```python
inverted_index = {}

for doc_id, text in documents.items():
    words = text.lower().replace('.', '').split()

    for word in words:
        if word not in inverted_index:
            inverted_index[word] = set()
        inverted_index[word].add(doc_id)

print("Inverted Index created successfully!")

print("\nDocuments containing 'python':", inverted_index.get('python', []))
print("Documents containing 'science':", inverted_index.get('science', []))
```

```python
def retrieve_documents(query, index):
    query_words = query.lower().split()

    if not query_words or query_words[0] not in index:
        return set()

    result_docs = index[query_words[0]]

    for word in query_words[1:]:
        if word in index:
            result_docs = result_docs.intersection(index[word])
        else:
            return set()

    return result_docs

search_query = "data science"
retrieved_ids = retrieve_documents(search_query, inverted_index)

print(f"Search Query: '{search_query}'")
print(f"Matching Document IDs: {retrieved_ids}\n")

print("Retrieved Documents:")
for doc_id in retrieved_ids:
    print(f"-> {documents[doc_id]}")
```
