```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer

nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab')
```

```python
text = "The quick brown foxes are jumping over the lazy dogs. They are very happy and excited!"
print("Original Text:\n", text)

words = word_tokenize(text.lower())
print("\nTokenized Words:\n", words)
```

```python
stop_words = set(stopwords.words('english'))

filtered_words = [word for word in words if word.isalnum() and word not in stop_words]

print("After Stop Word Removal:\n", filtered_words)
```

```python
stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(word) for word in filtered_words]

print("After Stemming:\n", stemmed_words)
```
